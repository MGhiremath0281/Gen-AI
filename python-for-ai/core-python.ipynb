## 1.1 Variables

A **variable** is a name that refers to a value stored in memory.

Python is **dynamically typed** — you don't declare the type, it is inferred
from the value you assign.

````python
name = "Nitish"
age = 22
temperature = 0.7

# check the type at runtime
print(type(name))
print(type(age))
print(type(temperature))
````

### Why it matters in AI

Almost every LLM call is just a few variables passed around:

````python
prompt = "Explain recursion"
model = "gpt-5"
temperature = 0.7
response = "Recursion is a function calling itself..."

print(f"model={model} | temperature={temperature}")
print(f"prompt={prompt}")
````

## 1.2 Data Types

A **data type** determines what kind of value a variable holds and what
operations are allowed on it.

| Type       | Example                  | Used for                    |
|------------|--------------------------|-----------------------------|
| `str`      | `"Nitish"`               | prompts, responses, text    |
| `int`      | `22`                     | counts, token limits        |
| `float`    | `0.7`                    | temperature, similarity     |
| `bool`     | `True`                   | flags, conditions           |
| `list`     | `["Python"]`             | messages, documents, chunks |
| `dict`     | `{"name": "Nitish"}`     | API request / response      |
| `NoneType` | `None`                   | "no value yet"              |

````python
name        = "Nitish"          # str
age         = 22                # int
temperature = 0.7               # float
is_active   = True              # bool
skills      = ["Python"]        # list
user        = {"name": "Nitish"}  # dict
value       = None              # NoneType

for v in (name, age, temperature, is_active, skills, user, value):
    print(f"{str(v):<20} -> {type(v).__name__}")
````

**The three you will use constantly:** `str` (prompts and responses),
`list` (collections of messages or documents), `dict` (API payloads).

## 1.6 Lists

A **list** is an ordered, mutable collection of values.

````python
languages = ["Python", "Java", "C++"]

# indexing and slicing
print(languages[0])     # first
print(languages[-1])    # last
print(languages[0:2])   # first two
print(len(languages))   # length
````

````python
languages = ["Python", "Java", "C++"]

languages.append("JavaScript")        # 1. add at the end
print("append   ->", languages)

languages.insert(1, "C")              # 2. add at an index
print("insert   ->", languages)

languages.extend(["Go", "Rust"])      # 3. add many
print("extend   ->", languages)

languages.remove("Java")              # 4. remove by value
print("remove   ->", languages)

languages.pop(2)                      # 5. remove by index
print("pop      ->", languages)

print("index    ->", languages.index("Python"))   # 6. position of a value
print("count    ->", languages.count("Python"))   # 7. occurrences

languages.sort()                      # 8. sort in place
print("sort     ->", languages)

languages.reverse()                   # 9. reverse in place
print("reverse  ->", languages)

copy_of = languages.copy()            # 10. shallow copy
print("copy     ->", copy_of)

languages.clear()                     # 11. empty the list
print("clear    ->", languages)
````

### List comprehension — a compact `for` loop that builds a list

````python
documents = ["doc1", "doc2", "doc3"]

upper = [d.upper() for d in documents]
print(upper)
````

## 1.7 Dictionaries

A **dictionary** stores data as **key → value** pairs. This is the shape of
almost every API request and response you will touch.

````python
student = {
    "name": "Muktananda",
    "age": 21,
    "course": "CSE",
}

print(student["name"])       # direct access (KeyError if missing)
print(student.get("email"))  # safe access (None if missing)
````

````python
student = {"name": "Muktananda", "age": 21, "course": "CSE"}

print("keys       ->", student.keys())       # 1. all keys
print("values     ->", student.values())     # 2. all values
print("items      ->", student.items())      # 3. key-value pairs
print("get        ->", student.get("name"))  # 4. safe lookup

student.update({"age": 22})                  # 5. add / update
print("update     ->", student)

student.pop("age")                           # 6. remove a key
print("pop        ->", student)

student.popitem()                            # 7. remove the last pair
print("popitem    ->", student)

student.setdefault("city", "Kalaburagi")     # 8. get, or insert if missing
print("setdefault ->", student)

copy_of = student.copy()                     # 9. shallow copy
print("copy       ->", copy_of)

student.clear()                              # 10. empty the dict
print("clear      ->", student)
````

### Looping over a dictionary

````python
config = {"model": "gpt-5", "temperature": 0.7, "max_tokens": 500}

for key, value in config.items():
    print(f"{key:<12} = {value}")
````

## 1.8 File Handling

File handling lets a program read data from and write data to files.

The standard pattern uses `with`, which closes the file automatically even if
an error happens.

| Mode  | Meaning                          |
|-------|----------------------------------|
| `"r"` | read (file must exist)           |
| `"w"` | write (overwrites existing file) |
| `"a"` | append to the end                |

````python
# writing
with open("data.txt", "w") as file:
    file.write("Hello AI\n")
    file.write("Second line\n")

print("written")
````

````python
# reading the whole file
with open("data.txt", "r") as file:
    content = file.read()

print(content)
````

````python
# reading line by line
with open("data.txt", "r") as file:
    for line in file:
        print(line.strip())
````

## 1.9 JSON

**JSON** (JavaScript Object Notation) is a text format for exchanging
structured data. Every LLM API speaks JSON.

```json
{
    "name": "Nitish",
    "age": 22,
    "skills": ["Python", "Java"]
}
```

| Function       | Direction                  |
|----------------|----------------------------|
| `json.loads()` | JSON string → Python object |
| `json.dumps()` | Python object → JSON string |
| `json.load()`  | JSON file → Python object   |
| `json.dump()`  | Python object → JSON file   |

````python
import json

# json.loads() : JSON string -> Python dict
data = '{"name": "Nitish", "age": 22}'
user = json.loads(data)

print(user["name"])
print(type(user))
````

````python
import json

# json.dumps() : Python dict -> JSON string
user = {"name": "Nitish", "age": 22, "skills": ["Python", "Java"]}
data = json.dumps(user, indent=2)

print(data)
print(type(data))
````

## 2.1 Classes

A **class** is a blueprint for creating objects. It defines what something
*has* (attributes) and what it *can do* (methods).

````python
class Chatbot:
    pass

# The class itself is not a particular chatbot yet -
# it only describes what a chatbot will look like.
print(Chatbot)
````

## 2.2 Objects

An **object** is an actual instance created from a class.

```text
Chatbot  -> class
bot1     -> object
bot2     -> object
```

````python
class Chatbot:
    pass

bot1 = Chatbot()
bot2 = Chatbot()

print(bot1)
print(bot2)
print(bot1 is bot2)   # two separate objects from the same class
````

## 2.3 `__init__()`

`__init__()` is a special method that runs **automatically** when an object is
created. It is used to initialise the object's data.

````python
class Chatbot:
    def __init__(self, name):
        self.name = name

bot = Chatbot("MyBot")
print(bot.name)
````

## 2.4 `self`

`self` refers to **the current object**.

Inside `__init__`, `self.name = name` means:
*"this particular bot's name is whatever was passed in."*

````python
class Chatbot:
    def __init__(self, name):
        self.name = name

bot_a = Chatbot("MyBot")
bot_b = Chatbot("OtherBot")

print(bot_a.name)   # each object keeps its own data
print(bot_b.name)
````

## 2.5 Attributes

An **attribute** is data stored inside an object.

````python
class Chatbot:
    def __init__(self, name, model):
        self.name = name
        self.model = model

bot = Chatbot("Gimi", "1.2")

print(bot.name)
print(bot.model)
````

## 2.6 Methods

A **method** is a function defined inside a class. Its first parameter is
always `self`.

````python
class Chatbot:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print("Hello from", self.name)

bot = Chatbot("GPT-5")
bot.greet()      # call it directly - do NOT wrap in print(),
                 # greet() returns None and would print "None" as well
````

### One class, many objects — each with its own state

````python
bot1 = Chatbot("gpt-5")
bot2 = Chatbot("gpt-4")

bot1.greet()
bot2.greet()

print(bot1.name, "|", bot2.name)
````

### A method that returns a value instead of printing

````python
class Chatbot:
    def __init__(self, name, temperature=0.7):
        self.name = name
        self.temperature = temperature

    def ask(self, question: str) -> str:
        return f"[{self.name} @ temp={self.temperature}] answering: {question}"

bot = Chatbot("GPT-5", temperature=0.2)
print(bot.ask("What is recursion?"))
````

## 2.7 Inheritance

**Inheritance** lets one class reuse and extend another class.

```text
Chatbot            (parent / base class)
   ↑
VoiceChatbot       (child class - gets ask(), adds speak())
```

````python
class Chatbot:
    def ask(self, question):
        print("Answering:", question)


class VoiceChatbot(Chatbot):       # inherits everything from Chatbot
    def speak(self):
        print("Speaking...")


bot = VoiceChatbot()
bot.ask("What is AI?")    # inherited method
bot.speak()               # its own method
````

### Overriding and `super()`

````python
class Chatbot:
    def __init__(self, name):
        self.name = name

    def ask(self, question):
        return f"{self.name}: {question}"


class VoiceChatbot(Chatbot):
    def __init__(self, name, voice):
        super().__init__(name)     # run the parent's __init__
        self.voice = voice

    def ask(self, question):       # override the parent's version
        base = super().ask(question)
        return f"{base} (spoken in {self.voice})"


bot = VoiceChatbot("GPT-5", voice="calm")
print(bot.ask("What is AI?"))
````

# 3. Async / Await
---

## 3.1 Why Asynchronous Programming Is Needed

**Asynchronous programming** lets a program start a long-running task (such as
an API call) and keep doing other work while waiting for it to finish.

This matters for LLM applications because the bottleneck is almost always
**network latency**, not CPU.

```text
Your program
    ↓
Send request to LLM
    ↓
WAITING...   ← network round trip (this is the expensive part)
    ↓
Receive response
```

## 3.2 Synchronous vs Asynchronous

**Synchronous:** tasks run one after another. The next task waits for the
current one to finish.

```text
Task 1 → wait → finish
                  ↓
              Task 2 → wait → finish
                              ↓
                          Task 3
```

````python
import time

def task(name):
    print("Starting", name)
    time.sleep(2)              # blocks the whole program
    print("Finished", name)

start = time.time()
task("A")
task("B")
print(f"Total: {time.time() - start:.1f}s")   # ~4 seconds
````

## 3.3 Asynchronous Execution

With async, tasks can wait **without blocking** the rest of the program.

Instead of sequential execution:

```text
Task A -> wait -> finish -> Task B -> wait -> finish
```

asynchronous execution overlaps the waiting:

```text
Task A -> waiting ─────────> finish
Task B -> waiting ─────────> finish
```

This is exactly what you want when firing several independent API calls.

## 3.4 `async def`

`async def` defines an **asynchronous function**, also called a *coroutine
function*.

Calling it does **not** run it — it returns a coroutine object that must be
awaited or run by an event loop.

````python
# normal function: calling it runs it
def hello():
    print("Hello")

hello()
````

````python
# async function: calling it returns a coroutine, nothing is printed yet
async def hello():
    print("Hello")

coro = hello()
print(coro)

coro.close()   # close it so Python does not warn about a never-awaited coroutine
````

The `async` keyword tells Python: *this function can perform asynchronous
operations.*

## 3.5 `await`

`await` tells Python to wait for an asynchronous operation to finish
**without blocking other async tasks**.

`await asyncio.sleep(2)` pauses only this coroutine for 2 seconds and lets
other coroutines run. `time.sleep(2)` blocks the entire thread — very
different.

````python
import asyncio

async def task():
    print("Starting")
    await asyncio.sleep(2)   # non-blocking pause
    print("Finished")
````

## 3.6 `asyncio`

`asyncio` is Python's built-in library for writing and running asynchronous
programs. The three pieces used most often:

| Function            | Purpose                                     |
|---------------------|---------------------------------------------|
| `asyncio.run()`     | run a coroutine from normal (sync) code     |
| `asyncio.sleep()`   | non-blocking pause                          |
| `asyncio.gather()`  | run several coroutines concurrently         |

## 3.7 Running an Async Function

```text
async def
   ↓
creates an async function (coroutine)
   ↓
asyncio.run()
   ↓
actually runs it
```

> ⚠️ **Inside Jupyter:** an event loop is already running, so `asyncio.run()`
> raises `RuntimeError: asyncio.run() cannot be called from a running event
> loop`. In a notebook, just use `await ...` at the top level of a cell.
> In a normal `.py` script, use `asyncio.run(main())`.

````python
import asyncio

async def hello():
    print("Hello")

# In a notebook: top-level await works directly
await hello()

# In a .py script you would instead write:
# asyncio.run(hello())
````

## 3.8 `await` in Action

A task that takes some time:

````python
import asyncio, time

async def task():
    print("Starting")
    await asyncio.sleep(2)
    print("Finished")

start = time.time()
await task()
print(f"Total: {time.time() - start:.1f}s")
````

## 3.9 Concurrent Execution

Concurrency is what makes AI applications fast. Querying several LLMs,
fetching embeddings, hitting a vector database — all of these are I/O bound.

Three independent calls, 2 seconds each:

**Synchronous — 6 seconds**

```text
API 1: ██████████ (2s)
API 2:           ██████████ (2s)
API 3:                     ██████████ (2s)
-------------------------------------------
Total ≈ 6 seconds
```

**Concurrent — 2 seconds**

```text
API 1: ██████████ (2s)
API 2: ██████████ (2s)
API 3: ██████████ (2s)
-------------------------------------------
Total ≈ 2 seconds
```

**Key takeaway:** concurrency overlaps the waiting, so total time ≈ the
slowest single call.

## 3.10 `asyncio.gather()`

`asyncio.gather()` runs multiple coroutines concurrently and waits for all of
them to finish, returning their results **in the order they were passed in**.

````python
import asyncio

async def task(name):
    print("Starting", name)
    await asyncio.sleep(2)
    print("Finished", name)
    return name
````

````python
import time

async def main():
    results = await asyncio.gather(
        task("A"),
        task("B"),
        task("C"),
    )
    print("Results:", results)

start = time.time()
await main()
print(f"Total: {time.time() - start:.1f}s")   # ~2s, not ~6s
````

## 3.11 Multiple LLM / API Calls Concurrently

When requests do not depend on each other, dispatch them all at once and
collect the results when they are done.

```text
             ┌──→ LLM 1 ──→ Response 1
             │
Question ────┼──→ LLM 2 ──→ Response 2
             │
             └──→ LLM 3 ──→ Response 3
                         │
                      gather
                         │
                         ▼
                   All responses
```

````python
import asyncio, time

# Simulated LLM call - in a real app this would be an HTTP request
async def call_llm(question: str) -> str:
    await asyncio.sleep(2)
    return f"Answer to: {question}"
````

````python
# Sequential (blocking): total = t1 + t2 + t3
start = time.time()

response1 = await call_llm("Question 1")
response2 = await call_llm("Question 2")
response3 = await call_llm("Question 3")

print(response1, "|", response2, "|", response3)
print(f"Sequential total: {time.time() - start:.1f}s")
````

````python
# Concurrent (non-blocking): total ~= slowest single call
start = time.time()

responses = await asyncio.gather(
    call_llm("Question 1"),
    call_llm("Question 2"),
    call_llm("Question 3"),
)

print(responses)
print(f"Concurrent total: {time.time() - start:.1f}s")
````

# 4. Pydantic
---

## 4.1 Why Data Validation Is Needed

**Data validation** means checking that incoming data matches the expected
structure, type, and constraints *before* your application uses it.

### The problem

Your app expects `name → str` and `age → int`, but receives:

```python
user = {
    "name": "Nitish",
    "age": "twenty two"     # ← wrong type
}
```

Nothing fails immediately. It fails much later, deep inside some function that
tries to do arithmetic on `age` — a confusing bug far from its cause.

### The solution

Validate at the **boundary** of the application: coerce the data into the
expected types, or reject it immediately with a clear error.

## 4.2 What Is Pydantic?

**Pydantic** is a data validation and settings-management library. It uses
ordinary Python type annotations to define schemas, then parses and validates
data against them at runtime.

```bash
pip install pydantic
```

````python
# Install if needed (uncomment)
# %pip install pydantic

import pydantic
print("pydantic", pydantic.VERSION)
````

## 4.3 `BaseModel`

`BaseModel` is the base class for all Pydantic schemas. Inheriting from it
defines the structure your data must follow.

````python
from pydantic import BaseModel

class User(BaseModel):
    name: str
    age: int

print(User.model_fields)
````

This model enforces two guarantees:

- `name` must be a valid `str`
- `age` must be a valid `int`

## 4.4 Defining Fields

Fields are class attributes with explicit type hints.

````python
from pydantic import BaseModel

class User(BaseModel):
    name: str   # field: name | expected type: str
    age: int    # field: age  | expected type: int

user = User(name="Nitish", age=22)

# attributes are accessed with normal dot notation
print(user.name)
print(user.age)
````

## 4.5 Type Annotations

You already saw type annotations with functions:

```python
def add(a: int, b: int) -> int:
    return a + b
```

Pydantic uses the exact same syntax to decide what each field expects:

```python
class User(BaseModel):
    name: str
    age: int
```

The difference: in a plain function the hints are only documentation, while
Pydantic actually **enforces** them.

## 4.6 Creating Pydantic Models

A slightly bigger model:

````python
from pydantic import BaseModel

class User(BaseModel):
    name: str
    age: int
    email: str

user = User(
    name="Nitish",
    age=22,
    email="nitish@example.com",
)

print(user)
````

## 4.7 Data Validation

### Valid input

```python
user = User(name="Nitish", age=22)   # ✅
```

### Invalid input

```python
user = User(name="Nitish", age="hello")   # ❌ ValidationError
```

```text
age
 │
 ├── Expected: int
 ├── Received: "hello"
 │
 └── ❌ ValidationError raised
```

````python
from pydantic import BaseModel, ValidationError

class User(BaseModel):
    name: str
    age: int

try:
    User(name="Nitish", age="hello")
except ValidationError as e:
    print(e)
````

### Strict validation vs type coercion

Pydantic is a **parsing** library, not only a type checker. If a value can be
safely converted to the declared type, it converts it.

> **Key rule:** Pydantic does not mean *"reject anything whose type is not an
> exact match."* It means *"parse the data into the declared schema, and raise
> an error only when safe parsing is impossible."*

````python
user = User(name="Nitish", age="22")   # "22" can be parsed into an int

print(user.age)         # 22
print(type(user.age))   # <class 'int'>
````

## 4.8 Nested Models

A **nested model** is a Pydantic model used as the field type of another
model. This is how you validate hierarchical JSON cleanly.

```text
User
 ├── name: str
 ├── age: int
 └── address: Address
       ├── city: str
       └── country: str
```

````python
from pydantic import BaseModel

class Address(BaseModel):
    city: str
    country: str

class User(BaseModel):
    name: str
    age: int
    address: Address

user = User(
    name="Nitish",
    age=22,
    address={"city": "Bengaluru", "country": "India"},   # plain dict is parsed
)

print(user.address.city)
print(type(user.address))
print(user)
````

**Key takeaway:** nested models let you validate and access deeply nested API
or LLM payloads with dot notation instead of chained `["key"]` lookups.

## 4.9 Optional Fields

An **optional field** is not required when creating the object and may be
`None`.

```python
# Python 3.10+
phone: str | None = None

# older syntax
from typing import Optional
phone: Optional[str] = None
```

> ⚠️ Always give it `= None`. Writing `phone: str | None` **without** a default
> still makes the key required — you would have to pass `phone=None` explicitly.

````python
from pydantic import BaseModel

class User(BaseModel):
    name: str
    age: int
    phone: str | None = None

# omitting the optional field
u1 = User(name="Nitish", age=22)
print(u1.phone)     # None

# providing it
u2 = User(name="Nitish", age=22, phone="9876543210")
print(u2.phone)
````

## 4.10 Default Fields

A **default field** has a fallback value that Pydantic assigns when nothing is
passed. Good for configuration flags and sensible baselines.

````python
from pydantic import BaseModel

class User(BaseModel):
    name: str
    active: bool = True    # default

user = User(name="Nitish")
print(user.active)         # True

inactive_user = User(name="Nitish", active=False)
print(inactive_user.active)   # False
````

## 4.11 Serialization: Model ↔ Dict ↔ JSON

```text
                 model_dump()
Pydantic Model ───────────────→ Dictionary
      │
      │ model_dump_json()
      ↓
    JSON string


Dictionary  ──→ model_validate()      ──→ Pydantic Model
JSON string ──→ model_validate_json() ──→ Pydantic Model
```

| Method                   | Direction               |
|--------------------------|-------------------------|
| `.model_dump()`          | model → dict            |
| `.model_dump_json()`     | model → JSON string     |
| `.model_validate()`      | dict → model            |
| `.model_validate_json()` | JSON string → model     |

> **Note:** these are the Pydantic **v2** methods. They replace v1's
> `.dict()`, `.json()` and `.parse_obj()`.

````python
from pydantic import BaseModel

class User(BaseModel):
    name: str
    age: int

user = User(name="Nitish", age=22)

# model -> dict
data = user.model_dump()
print(data, type(data))

# model -> JSON string
json_data = user.model_dump_json()
print(json_data, type(json_data))
````

````python
# dict -> model
data = {"name": "Nitish", "age": 22}
user = User.model_validate(data)
print(user.name)

# JSON string -> model
json_data = '{"name": "Nitish", "age": 22}'
user = User.model_validate_json(json_data)
print(user.age)
````

## 4.12 Pydantic in LLM Applications

An LLM naturally returns unstructured text:

```text
"Sure! Here are the user details you requested:
Name: John
Age: 25
Skills: Python, Java"
```

Parsing that with regex or string splitting is brittle. Instead, declare the
shape you want and let Pydantic enforce it:

```python
class Person(BaseModel):
    name: str
    age: int
    skills: list[str]
```

That model becomes a **type-safe contract** for data entering and leaving the
LLM pipeline.

### 4.13 Structured Input

**Structured input** means sending prompts and parameters as validated schemas
rather than loose dictionaries.

````python
from pydantic import BaseModel

# Loose and unvalidated - a typo in a key fails silently
payload = {"question": "Explain recursion", "difficulty": "medium"}

# Typed and validated
class Question(BaseModel):
    question: str
    difficulty: str

query = Question(question="Explain recursion", difficulty="medium")
print(query)
````

### 4.14 Structured Output

**Structured output** means asking the LLM to return data matching a schema
(usually JSON) and parsing it straight into a model.

```text
       Prompt / Request
              │
              ▼
           LLM Engine
              │
              ▼
   Raw JSON / structured data
              │
              ▼
   Pydantic validation (.model_validate_json())
              │
              ▼
     Typed Person object
```

````python
from pydantic import BaseModel

class Person(BaseModel):
    name: str
    age: int
    skills: list[str]

# pretend this string came back from an LLM
llm_output = '{"name": "John", "age": 25, "skills": ["Python", "Java"]}'

person = Person.model_validate_json(llm_output)

print(person.name)
print(person.skills)
print(type(person))
````

**Key takeaway:** structured output turns unpredictable text generation into
reliable, typed objects. This is why Pydantic sits underneath almost every
modern LLM framework.

# 5. Environment & Dependency Management
---

This is very important for AI projects, because they depend on many external
libraries — `pydantic`, `openai`, `langchain`, `python-dotenv`, and so on.

> **The main idea:** keep each project's Python environment, packages, and
> secrets isolated and reproducible.

## 5.1 Virtual Environments

A **virtual environment** is an isolated Python environment for one project.

```text
Your Computer
│
├── Project A
│   └── .venv
│       ├── Python
│       └── Packages
│
└── Project B
    └── .venv
        ├── Python
        └── Packages
```

### Why

```text
Project A → pydantic 2.x
Project B → pydantic 1.x
```

Without isolation these conflict. With a `.venv` each project keeps its own
package versions and they never interfere.

## 5.2 `venv`

Python ships with a built-in module called `venv`.

Inside your project folder:

```powershell
python -m venv .venv
```

Which gives you:

```text
MyProject/
│
├── .venv/
├── main.py
└── ...
```

`.venv` is just the conventional name — any name works.

## 5.3 Activating a Virtual Environment

**Windows PowerShell**

```powershell
.\.venv\Scripts\Activate.ps1
```

**macOS / Linux**

```bash
source .venv/bin/activate
```

After activation the prompt shows:

```text
(.venv) PS C:\Projects\MyProject>
```

That `(.venv)` means `python` and `pip` now point at this project's
environment.

**Deactivate** when finished:

```powershell
deactivate
```

## 5.4 Installing Packages with `pip`

`pip` is Python's package installer:

```powershell
pip install pydantic
```

Prefer this longer form:

```powershell
python -m pip install <package>
```

because it makes explicit that you are using the `pip` belonging to *that*
Python interpreter.

```powershell
python -m pip install pydantic
python -m pip install python-dotenv
```

## 5.5 `.env` Files

Suppose your application needs an API key:

```python
api_key = "sk-xxxxxxxxxxxxxxxx"   # ❌ never do this
```

Your source code may end up on GitHub — the key would go with it.

Instead put it in a `.env` file:

```text
OPENAI_API_KEY=your_api_key_here
```

```text
.env
 ↓
API Key
 ↓
Python application
 ↓
OpenAI API
```

## 5.6 Environment Variables

An **environment variable** is a value stored outside your source code that
the program can read at runtime.

```text
OPENAI_API_KEY=abc123
```

````python
import os

# reads from the environment; returns None if not set
api_key = os.getenv("OPENAI_API_KEY")
print(api_key)

# with a fallback default
model = os.getenv("MODEL_NAME", "gpt-5")
print(model)
````

## 5.7 `python-dotenv`

Python does not read `.env` files by itself. `python-dotenv` does.

```powershell
python -m pip install python-dotenv
```

`.env`

```text
OPENAI_API_KEY=abc123
MODEL_NAME=gpt-5
```

```python
import os
from dotenv import load_dotenv

load_dotenv()

api_key = os.getenv("OPENAI_API_KEY")
model = os.getenv("MODEL_NAME")
```

What happens:

```text
.env
 │
 │ load_dotenv()
 ↓
Environment Variables
 │
 │ os.getenv()
 ↓
Python Program
```

````python
# Runnable demo: write a .env, then load it
# (in a real project the .env file already exists and is git-ignored)

with open(".env", "w") as f:
    f.write("OPENAI_API_KEY=abc123\n")
    f.write("MODEL_NAME=gpt-5\n")

try:
    from dotenv import load_dotenv
    import os

    load_dotenv()
    print(os.getenv("OPENAI_API_KEY"))
    print(os.getenv("MODEL_NAME"))
except ImportError:
    print("python-dotenv not installed - run: python -m pip install python-dotenv")
````

## 5.9 `requirements.txt`

When someone else downloads your project, they need to know which packages it
depends on. That is what `requirements.txt` records:

```text
pydantic
python-dotenv
openai
```

## 5.10 Installing from `requirements.txt`

```text
AIProject/
│
├── main.py
├── requirements.txt
└── .env
```

```powershell
python -m venv .venv
.\.venv\Scripts\Activate.ps1
python -m pip install -r requirements.txt
```

`-r` means *read the packages from this file and install them*.

```text
requirements.txt
       ↓
pip install -r
       ↓
All required packages
       ↓
.venv
```

## 5.11 Pinning Package Versions

Today your project uses `pydantic 2.11`. Six months later someone installs
`pydantic 3.x`, the behaviour changes, and the project breaks.

So pin the exact version:

```text
package==version
```

```text
pydantic==2.11.7
python-dotenv==1.1.1
```

### Generating `requirements.txt` automatically

```powershell
python -m pip freeze > requirements.txt
```

This writes every package installed in the current environment, with versions.
Another developer then runs:

```powershell
python -m pip install -r requirements.txt
```
