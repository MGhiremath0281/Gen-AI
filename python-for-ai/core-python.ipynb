1.6 Lists
Definition

A list is an ordered, mutable collection of values.

````python
languages = ["Python", "Java", "C++"]

# 1. append() - Add element at the end
languages.append("JavaScript")
print(languages)


# 2. insert() - Add element at a specific index
languages.insert(1, "C")
print(languages)


# 3. extend() - Add multiple elements
languages.extend(["Go", "Rust"])
print(languages)


# 4. remove() - Remove a specific element
languages.remove("Java")
print(languages)


# 5. pop() - Remove element using index
languages.pop(2)
print(languages)


# 6. index() - Find index of an element
print(languages.index("Python"))


# 7. count() - Count occurrences
print(languages.count("Python"))


# 8. sort() - Sort the list
languages.sort()
print(languages)


# 9. reverse() - Reverse the list
languages.reverse()
print(languages)


# 10. copy() - Copy the list
new_languages = languages.copy()
print(new_languages)


# 11. clear() - Remove all elements
languages.clear()
print(languages)
````

````python
# List + Loop
documents = ["doc1", "doc2", "doc3"]

for document in documents:
    print(document)
````

1.7 Dictionaries
Definition

A dictionary stores data as key-value pairs.

````python
student = {
    "name": "Muktananda",
    "age": 21,
    "course": "CSE"
}

# 1. keys() - Get all keys
print(student.keys())


# 2. values() - Get all values
print(student.values())


# 3. items() - Get key-value pairs
print(student.items())


# 4. get() - Get value using key
print(student.get("name"))


# 5. update() - Add or update key-value pairs
student.update({"age": 22})
print(student)


# 6. pop() - Remove a key and return its value
student.pop("age")
print(student)


# 7. popitem() - Remove the last key-value pair
student.popitem()
print(student)


# 8. setdefault() - Get value; if key doesn't exist, add it
student.setdefault("city", "Kalaburagi")
print(student)


# 9. copy() - Create a copy
new_student = student.copy()
print(new_student)


# 10. clear() - Remove everything
student.clear()
print(student)
````

1.9 JSON
Definition

JSON (JavaScript Object Notation) is a text-based format used to represent and exchange structured data.
``` JSON
{
    "name": "Nitish",
    "age": 22,
    "skills": ["Python", "Java"]
}
```

````python
# json.loads() : - json.loads() converts a JSON string into a Python object.
import json

data = '{"name": "Nitish", "age": 22}'

user = json.loads(data)

print(user["name"])
````

````python
# json.dumps() : json.dumps() converts a Python object into a JSON string.
import json

user = {
    "name": "Nitish",
    "age": 22
}

data = json.dumps(user)

print(data)
````

2.9 Basic Inheritance
Definition

Inheritance allows one class to reuse and extend another class..

````python
class Chatbot:

    def ask(self, question):
        print("Answering:", question)


class VoiceChatbot(Chatbot):

    def speak(self):
        print("Speaking...")
````

````python
bot = VoiceChatbot()

bot.ask("What is AI?")
bot.speak()
````
