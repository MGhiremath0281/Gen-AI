# Part 1: Basics

## Key facts
- Python is **interpreted** and **dynamically typed** (no need to declare type).
- **Indentation** (4 spaces) defines blocks. No `{}`.
- Everything is an **object**.
- `#` is a comment. Triple quotes `""" """` are used for docstrings / multi-line strings.
- Naming: `snake_case` for variables/functions, `PascalCase` for classes, `UPPER_CASE` for constants.

````python
print("Hello, Python!")
print("a", "b", "c", sep="-", end="!\n")   # sep and end control output

name = "Muk"
age = 21
price = 99.5
is_student = True
nothing = None

print(name, age, price, is_student, nothing)
print(type(name), type(age), type(price), type(is_student), type(nothing))
````

## Operators
- Arithmetic: `+ - * / // % **` (`/` always gives float, `//` is floor division)
- Comparison: `== != < > <= >=` (can chain: `1 < x < 10`)
- Logical: `and or not` (short-circuit)
- Identity: `is`, `is not` (same object). Membership: `in`, `not in`
- Assignment: `+= -= *= //=` etc.
- Bitwise: `& | ^ ~ << >>`

````python
print(7 / 2, 7 // 2, 7 % 2, 2 ** 10, -7 // 2)
print(1 < 5 < 10, not True, True and False, True or False)

a = [1, 2]; b = [1, 2]; c = a
print(a == b, a is b, a is c)      # == compares value, is compares identity
print(3 in [1, 2, 3], "x" not in "abc")

print(5 & 3, 5 | 3, 5 ^ 3, 1 << 3)   # bitwise

# short-circuit trick: default value
user_input = ""
print(user_input or "default")

# walrus operator (3.8+)
if (n := len("hello")) > 3:
    print("long word, length", n)
````

# Part 2: Strings

- Strings are **immutable** (you cannot change a char in place, you create a new string).
- Indexing starts at `0`, negative index counts from the end.
- Slicing: `s[start:stop:step]` (stop is excluded).

````python
s = "Python Programming"
print(s[0], s[-1], s[0:6], s[7:], s[::-1], s[::2])
print(len(s), s.upper(), s.lower(), s.title(), s.swapcase())
print(s.find("gram"), s.index("P"), s.count("m"), s.find("zzz"))  # find gives -1 if missing
print(s.startswith("Py"), s.endswith("ing"), "Pro" in s)
print(s.replace("Python", "Java"))
````

````python
# strip, split, join: used all the time
raw = "   hello world, python  "
print(repr(raw.strip()), repr(raw.lstrip()), repr(raw.rstrip()))

words = "a,b,c,d".split(",")
print(words, "-".join(words))
print("one  two   three".split())          # split() with no arg splits on any whitespace

# checks
print("abc".isalpha(), "123".isdigit(), "abc123".isalnum(), "  ".isspace(), "Abc".istitle())

# padding / alignment
print("7".zfill(3), "hi".center(10, "*"), "hi".ljust(6, ".") + "|", "hi".rjust(6, ".") )

# multi-line and raw strings
text = """line1
line2"""
print(text.splitlines())
print(r"C:\new\table")       # raw string: backslashes are kept as is
````

````python
# common string tasks
s = "A man a plan a canal Panama"
clean = "".join(ch.lower() for ch in s if ch.isalnum())
print(clean == clean[::-1])                      # palindrome

print("".join(sorted("listen")) == "".join(sorted("silent")))   # anagram

sentence = "the quick brown fox jumps over the lazy dog the end"
freq = {}
for w in sentence.split():
    freq[w] = freq.get(w, 0) + 1
print(freq)

print(ord("a"), chr(98))                         # char <-> ascii
````

## List

````python
nums = [5, 3, 8, 1]
nums.append(10)             # add at end
nums.insert(0, 99)          # add at index
nums.extend([7, 7])         # add many
print(nums)

nums.remove(7)              # removes first 7
last = nums.pop()           # removes & returns last
print(last, nums)
del nums[0]
print(nums)

nums.sort()                 # in place
print(nums, sorted(nums, reverse=True))   # sorted() returns a new list
print(nums.index(8), nums.count(3), len(nums), min(nums), max(nums), sum(nums))
nums.reverse(); print(nums)
````

````python
# slicing, copy trap, 2D list
a = [1, 2, 3]
b = a            # same object!
b.append(4)
print(a)         # a also changed

c = a.copy()     # shallow copy (also a[:] or list(a))
c.append(5)
print(a, c)

import copy
nested = [[1, 2], [3, 4]]
shallow = nested.copy()
deep = copy.deepcopy(nested)
nested[0][0] = 100
print(shallow, deep)   # shallow shares inner lists, deep does not

# correct way to make a 2D grid
grid = [[0] * 3 for _ in range(2)]
grid[0][0] = 1
print(grid)
# WRONG: [[0]*3]*2  -> both rows are the same list
````

## Tuple
Immutable, can be used as dict key, good for fixed data and returning multiple values.

````python
t = (1, 2, 3)
single = (5,)              # comma makes it a tuple
x, y, z = t                # unpacking
print(t[0], t[-1], single, x, y, z, t.count(2), t.index(3))
# t[0] = 10  -> TypeError

from collections import namedtuple
Point = namedtuple("Point", "x y")
p = Point(3, 4)
print(p, p.x, p.y)
````

## Set
Unique items, very fast membership test, supports math operations.

````python
s = {1, 2, 3, 3, 2}
print(s)
s.add(4); s.discard(10); s.remove(1)     # remove raises KeyError if missing, discard does not
print(s)

A, B = {1, 2, 3, 4}, {3, 4, 5}
print(A | B, A & B, A - B, A ^ B)         # union, intersection, difference, symmetric diff
print(A.issubset({1, 2, 3, 4, 5}), A.isdisjoint({9}))

# remove duplicates, keep order
data = [3, 1, 3, 2, 1]
print(list(dict.fromkeys(data)))
print(len(set(data)) != len(data))        # has duplicates?
````

## Dict
Key → value. Keys must be **hashable** (str, int, tuple... not list).

````python
d = {"name": "Muk", "age": 21}
d["city"] = "Bangalore"             # add / update
print(d["name"], d.get("phone"), d.get("phone", "NA"))   # get avoids KeyError
d.pop("age")
print(d, "name" in d)

for k, v in d.items():
    print(k, "->", v)
print(list(d.keys()), list(d.values()))

d.update({"age": 22, "lang": "python"})
print(d)

# setdefault / merge (3.9+)
d.setdefault("country", "India")
merged = {"a": 1} | {"b": 2}
print(merged)

# dict comprehension, sort by value
sq = {n: n * n for n in range(5)}
print(sq)
scores = {"a": 50, "b": 90, "c": 70}
print(sorted(scores.items(), key=lambda kv: kv[1], reverse=True))
print(max(scores, key=scores.get))
````

## Comprehensions (very important)
`[expr for item in iterable if condition]` works for list, set and dict.

````python
print([n * n for n in range(1, 6)])
print([n for n in range(20) if n % 3 == 0])
print([("even" if n % 2 == 0 else "odd") for n in range(4)])
print({c for c in "hello"})
print({w: len(w) for w in ["hi", "python"]})
print([x for row in [[1, 2], [3, 4]] for x in row])   # flatten 2D list
````

# Part 4: Control Flow

- `if / elif / else`, ternary: `a if cond else b`
- `for` loops over any iterable. `while` loops till condition false.
- `break` exits loop, `continue` skips to next iteration, loop `else` runs only if loop was **not** broken.
- Useful: `range`, `enumerate`, `zip`, `reversed`, `sorted`, `any`, `all`.

````python
marks = 72
if marks >= 90:
    grade = "A"
elif marks >= 70:
    grade = "B"
else:
    grade = "C"
print(grade, "pass" if marks >= 40 else "fail")

for i in range(1, 10, 3):          # start, stop, step
    print(i, end=" ")
print()

for idx, ch in enumerate("abc", start=1):
    print(idx, ch)

names, ages = ["a", "b", "c"], [20, 21, 22]
for n, a in zip(names, ages):
    print(n, a)
print(dict(zip(names, ages)))
````

````python
# while, break, continue, loop-else
n = 0
while True:
    n += 1
    if n % 2 == 0:
        continue
    if n > 7:
        break
    print(n, end=" ")
print()

# prime check with for-else
num = 29
for i in range(2, int(num ** 0.5) + 1):
    if num % i == 0:
        print("not prime"); break
else:
    print("prime")             # runs only if no break

print(any([0, 0, 1]), all([1, 1, 0]))

# nested loop pattern
for i in range(1, 5):
    print("*" * i)
````

````python
# match-case (Python 3.10+), like switch
def http(code):
    match code:
        case 200:
            return "OK"
        case 404:
            return "Not Found"
        case 500 | 502:
            return "Server Error"
        case _:
            return "Unknown"
print(http(200), http(502), http(1))

# pattern match on structure
point = (0, 5)
match point:
    case (0, y):
        print("on y axis at", y)
    case (x, 0):
        print("on x axis at", x)
````

# Part 5: Functions

- Define with `def`. Without `return`, function gives `None`.
- **Parameters:** positional, default, `*args` (tuple of extras), `**kwargs` (dict of extras), keyword-only (after `*`).
- **Scope (LEGB):** Local → Enclosing → Global → Built-in.
- Functions are **first-class**: you can pass them around and return them.
- **Never** use a mutable default argument (`def f(x=[])`).

````python
def greet(name, greeting="Hello"):
    """Return a greeting string."""      # docstring
    return f"{greeting}, {name}!"

print(greet("Muk"), greet("Muk", greeting="Hi"))
print(greet.__doc__)

def stats(*args, **kwargs):
    print(args, kwargs)
stats(1, 2, 3, a=10, b=20)

def only_kw(a, *, b):          # b must be passed by keyword
    return a + b
print(only_kw(1, b=2))

def min_max(nums):
    return min(nums), max(nums)   # returns a tuple
lo, hi = min_max([4, 1, 9])
print(lo, hi)

nums = [1, 2, 3]
def add(a, b, c): return a + b + c
print(add(*nums))                  # unpack list as arguments
print(add(**{"a": 1, "b": 2, "c": 3}))   # unpack dict
````

````python
# mutable default trap
def bad(item, bucket=[]):
    bucket.append(item)
    return bucket
print(bad(1), bad(2))             # [1] then [1, 2]  <- surprise!

def good(item, bucket=None):
    if bucket is None:
        bucket = []
    bucket.append(item)
    return bucket
print(good(1), good(2))

# scope
x = "global"
def outer():
    x = "enclosing"
    def inner():
        nonlocal x
        x = "changed by inner"
    inner()
    return x
print(outer(), x)

counter = 0
def inc():
    global counter
    counter += 1
inc(); inc()
print(counter)
````

````python
# lambda, map, filter, reduce, sorted with key
sq = lambda n: n * n
print(sq(5))

nums = [1, 2, 3, 4, 5, 6]
print(list(map(lambda n: n * 2, nums)))
print(list(filter(lambda n: n % 2 == 0, nums)))

from functools import reduce
print(reduce(lambda a, b: a * b, nums))

people = [("a", 30), ("b", 20), ("c", 25)]
print(sorted(people, key=lambda p: p[1]))

# closures: function that remembers
def make_multiplier(n):
    def multiply(x):
        return x * n
    return multiply
double = make_multiplier(2)
print(double(8))
````

````python
# recursion: function calling itself. Needs a BASE CASE.
def factorial(n):
    if n <= 1:                 # base case
        return 1
    return n * factorial(n - 1)
print(factorial(5))

# memoization with lru_cache
from functools import lru_cache
@lru_cache(maxsize=None)
def fib(n):
    return n if n < 2 else fib(n - 1) + fib(n - 2)
print(fib(50))

import sys
print(sys.getrecursionlimit())     # default ~1000, deep recursion -> RecursionError
````

# Part 7: File Handling, CSV, JSON

- Always use `with open(...)`: it closes the file for you.
- Modes: `r` read, `w` write (overwrites), `a` append, `rb/wb` binary, add `+` for read+write.
- Prefer `pathlib.Path` over string paths.

````python
with open("demo.txt", "w") as f:
    f.write("line 1\n")
    f.writelines(["line 2\n", "line 3\n"])

with open("demo.txt") as f:
    print(f.read())

with open("demo.txt") as f:
    for line in f:                 # memory friendly, one line at a time
        print(line.strip())

with open("demo.txt", "a") as f:
    f.write("line 4\n")

from pathlib import Path
p = Path("demo.txt")
print(p.exists(), p.suffix, p.stem, p.parent.resolve().name)
print(p.read_text().splitlines())
````

````python
import csv, json

rows = [{"name": "A", "marks": 80}, {"name": "B", "marks": 65}]
with open("marks.csv", "w", newline="") as f:
    w = csv.DictWriter(f, fieldnames=["name", "marks"])
    w.writeheader()
    w.writerows(rows)

with open("marks.csv") as f:
    for row in csv.DictReader(f):
        print(row["name"], int(row["marks"]))

# JSON: dict <-> text
data = {"user": "muk", "skills": ["python", "java"], "active": True, "x": None}
text = json.dumps(data, indent=2)
print(text)
back = json.loads(text)
print(back["skills"])

with open("data.json", "w") as f:
    json.dump(data, f, indent=2)
with open("data.json") as f:
    print(json.load(f))
````

# Part 8: Exceptions

```
try:      risky code
except X: handle error X
else:     runs if no error
finally:  always runs (cleanup)
```
- `raise` to throw an error. Create custom errors by subclassing `Exception`.
- Catch **specific** exceptions. Avoid bare `except:`.
- Common: `ValueError, TypeError, KeyError, IndexError, ZeroDivisionError, FileNotFoundError, AttributeError, ImportError`.

````python
def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("cannot divide by zero")
        return None
    except TypeError as e:
        print("bad types:", e)
        return None
    else:
        print("no error")
        return result
    finally:
        print("done")

print(divide(10, 2))
print(divide(1, 0))
print(divide("a", 1))
````

````python
class InsufficientFunds(Exception):
    def __init__(self, balance, amount):
        super().__init__(f"Balance {balance}, tried to withdraw {amount}")
        self.balance, self.amount = balance, amount

def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientFunds(balance, amount)
    return balance - amount

try:
    withdraw(100, 500)
except InsufficientFunds as e:
    print("Error:", e)

# safe input loop pattern
def to_int(s):
    try:
        return int(s)
    except ValueError:
        return None
print(to_int("12"), to_int("x"))

try:
    open("missing.txt")
except FileNotFoundError as e:
    print("file problem:", e)

# exception chaining
try:
    try:
        int("x")
    except ValueError as e:
        raise RuntimeError("config invalid") from e
except RuntimeError as e:
    print(e, "| caused by:", repr(e.__cause__))
````

# Part 9: OOP (Object Oriented Programming)

- **Class** = blueprint, **object** = instance.
- `__init__` = constructor, `self` = current object.
- **Instance attribute** (per object) vs **class attribute** (shared).
- Methods: instance (`self`), `@classmethod` (`cls`), `@staticmethod` (no self/cls).
- **4 pillars:** Encapsulation, Inheritance, Polymorphism, Abstraction.
- Encapsulation in Python is by convention: `_protected`, `__private` (name mangled). Use `@property` for getters/setters.
- Dunder (magic) methods: `__str__, __repr__, __len__, __eq__, __lt__, __add__, __getitem__, __iter__, __call__`.

````python
class Student:
    school = "ABC University"            # class attribute

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def grade(self):                      # instance method
        return "A" if self.marks >= 80 else "B"

    @classmethod
    def from_string(cls, s):              # alternative constructor
        name, marks = s.split("-")
        return cls(name, int(marks))

    @staticmethod
    def is_valid(marks):                  # utility, no access to self
        return 0 <= marks <= 100

    def __str__(self):                    # for print()
        return f"{self.name} ({self.marks})"

    def __repr__(self):                   # for debugging
        return f"Student({self.name!r}, {self.marks})"

s1 = Student("Asha", 85)
s2 = Student.from_string("Ravi-72")
print(s1, repr(s2), s1.grade(), Student.is_valid(120), Student.school)
````

````python
# Encapsulation with property
class Account:
    def __init__(self, balance=0):
        self.__balance = balance          # private (name mangled)

    @property
    def balance(self):
        return self.__balance

    @balance.setter
    def balance(self, value):
        if value < 0:
            raise ValueError("balance cannot be negative")
        self.__balance = value

    def deposit(self, amt):
        self.balance += amt

acc = Account(100)
acc.deposit(50)
print(acc.balance)
# acc.balance = -5   -> ValueError
print(acc._Account__balance)             # still reachable: privacy is convention
````

````python
# Inheritance, super(), polymorphism
class Animal:
    def __init__(self, name):
        self.name = name
    def speak(self):
        return "..."

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed
    def speak(self):                      # method overriding
        return "Woof"

class Cat(Animal):
    def speak(self):
        return "Meow"

for a in [Dog("Rex", "Lab"), Cat("Tom"), Animal("Generic")]:
    print(a.name, a.speak())              # polymorphism: same call, different behaviour

print(isinstance(Dog("a", "b"), Animal), issubclass(Dog, Animal), Dog.__mro__)

# Abstraction with ABC
from abc import ABC, abstractmethod
class Shape(ABC):
    @abstractmethod
    def area(self): ...

class Circle(Shape):
    def __init__(self, r): self.r = r
    def area(self): return round(3.14159 * self.r ** 2, 2)

print(Circle(3).area())
# Shape() -> TypeError, cannot instantiate abstract class
````

````python
# Operator overloading and other dunder methods
class Vector:
    def __init__(self, x, y):
        self.x, self.y = x, y
    def __add__(self, o):  return Vector(self.x + o.x, self.y + o.y)
    def __eq__(self, o):   return (self.x, self.y) == (o.x, o.y)
    def __lt__(self, o):   return (self.x**2 + self.y**2) < (o.x**2 + o.y**2)
    def __len__(self):     return 2
    def __getitem__(self, i): return (self.x, self.y)[i]
    def __call__(self, k): return Vector(self.x * k, self.y * k)
    def __repr__(self):    return f"Vector({self.x}, {self.y})"

v = Vector(1, 2) + Vector(3, 4)
print(v, v == Vector(4, 6), len(v), v[0], v(2), Vector(1, 1) < v)

# Composition ("has-a") is often better than inheritance ("is-a")
class Engine:
    def start(self): return "engine on"
class Car:
    def __init__(self): self.engine = Engine()
    def start(self): return self.engine.start()
print(Car().start())
````

# Part 10: Iterators, Generators, Decorators, Context Managers

- **Iterable**: has `__iter__` (list, str, dict...). **Iterator**: has `__next__`, remembers position.
- **Generator**: function using `yield`. Lazy, saves memory, great for big data.
- **Decorator**: function that wraps another function to add behaviour.
- **Context manager**: object for `with` (`__enter__`, `__exit__`).

````python
it = iter([10, 20])
print(next(it), next(it))
# next(it) -> StopIteration

def countdown(n):
    while n > 0:
        yield n                 # pauses here, resumes on next call
        n -= 1

print(list(countdown(4)))

def fib_gen():
    a, b = 0, 1
    while True:                 # infinite but lazy
        yield a
        a, b = b, a + b

from itertools import islice
print(list(islice(fib_gen(), 10)))

gen_exp = (n * n for n in range(1_000_000))     # generator expression: no big list in memory
print(next(gen_exp), next(gen_exp))
print(sum(n for n in range(10)))
````

````python
# Decorators
import time
from functools import wraps

def timer(func):
    @wraps(func)                         # keeps original name and docstring
    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        result = func(*args, **kwargs)
        print(f"{func.__name__} took {time.perf_counter() - start:.5f}s")
        return result
    return wrapper

@timer
def slow_sum(n):
    """sum numbers"""
    return sum(range(n))

print(slow_sum(1_000_000), slow_sum.__name__)

# decorator with arguments
def repeat(times):
    def deco(func):
        @wraps(func)
        def wrapper(*a, **k):
            for _ in range(times):
                result = func(*a, **k)
            return result
        return wrapper
    return deco

@repeat(3)
def hello():
    print("hello")
hello()
````

````python
# Context managers
class Timer:
    def __enter__(self):
        self.t = time.perf_counter()
        return self
    def __exit__(self, exc_type, exc, tb):
        print(f"elapsed {time.perf_counter() - self.t:.4f}s")
        return False                    # do not swallow exceptions

with Timer():
    sum(range(500000))

from contextlib import contextmanager
@contextmanager
def tag(name):
    print(f"<{name}>")
    yield
    print(f"</{name}>")

with tag("div"):
    print("content")
````

````python
from dataclasses import dataclass, field
from enum import Enum
from typing import Optional

class Role(Enum):
    ADMIN = "admin"
    USER = "user"

@dataclass
class User:
    name: str
    age: int
    role: Role = Role.USER
    tags: list[str] = field(default_factory=list)    # mutable default done right

u = User("Muk", 21)
print(u, u == User("Muk", 21), Role("admin"), Role.ADMIN.name)

def find_user(users: list[User], name: str) -> Optional[User]:
    return next((x for x in users if x.name == name), None)
print(find_user([u], "Muk"), find_user([u], "zzz"))
````

# Part 13: Recursion + Interview Practice

**Recursion recipe:** 1) base case 2) smaller subproblem 3) combine result.
All solutions below are **recursive**.

````python
def reverse(s):
    return s if len(s) <= 1 else reverse(s[1:]) + s[0]

def is_palindrome(s):
    if len(s) <= 1: return True
    return s[0] == s[-1] and is_palindrome(s[1:-1])

def sum_digits(n):
    return n if n < 10 else n % 10 + sum_digits(n // 10)

def power(x, n):                    # fast power O(log n)
    if n == 0: return 1
    half = power(x, n // 2)
    return half * half if n % 2 == 0 else half * half * x

def gcd(a, b):
    return a if b == 0 else gcd(b, a % b)

def flatten(lst):
    out = []
    for x in lst:
        out += flatten(x) if isinstance(x, list) else [x]
    return out

print(reverse("python"), is_palindrome("level"), sum_digits(1234), power(2, 10), gcd(48, 18))
print(flatten([1, [2, [3, [4]], 5]]))
````

````python
def binary_search(arr, target, lo=0, hi=None):
    if hi is None: hi = len(arr) - 1
    if lo > hi: return -1
    mid = (lo + hi) // 2
    if arr[mid] == target: return mid
    if arr[mid] < target:  return binary_search(arr, target, mid + 1, hi)
    return binary_search(arr, target, lo, mid - 1)

def subsets(nums):
    if not nums: return [[]]
    rest = subsets(nums[1:])
    return rest + [[nums[0]] + r for r in rest]

def permutations(s):
    if len(s) <= 1: return [s]
    out = []
    for i, ch in enumerate(s):
        for p in permutations(s[:i] + s[i + 1:]):
            out.append(ch + p)
    return out

def merge_sort(a):
    if len(a) <= 1: return a
    m = len(a) // 2
    L, R = merge_sort(a[:m]), merge_sort(a[m:])
    def merge(L, R):
        if not L: return R
        if not R: return L
        return [L[0]] + merge(L[1:], R) if L[0] <= R[0] else [R[0]] + merge(L, R[1:])
    return merge(L, R)

print(binary_search([1, 3, 5, 7, 9], 7), subsets([1, 2, 3]))
print(permutations("abc"), merge_sort([5, 2, 9, 1, 5, 6]))
````

````python
# Common warm-up problems
def fizzbuzz(n):
    for i in range(1, n + 1):
        print("FizzBuzz" if i % 15 == 0 else "Fizz" if i % 3 == 0 else "Buzz" if i % 5 == 0 else i, end=" ")
    print()
fizzbuzz(15)

def two_sum(nums, target):           # dict trick, O(n)
    seen = {}
    for i, n in enumerate(nums):
        if target - n in seen:
            return [seen[target - n], i]
        seen[n] = i
print(two_sum([2, 7, 11, 15], 9))

def first_non_repeating(s):
    from collections import Counter
    c = Counter(s)
    return next((ch for ch in s if c[ch] == 1), None)
print(first_non_repeating("swiss"))

def word_freq(text):
    from collections import Counter
    return Counter(text.lower().split()).most_common(3)
print(word_freq("the cat and the hat and the bat"))
````

# Part 14: Interview Q&A (short answers)

**1. List vs tuple?** List is mutable, tuple is immutable (faster, hashable, usable as dict key).

**2. `==` vs `is`?** `==` compares values, `is` compares object identity.

**3. Mutable vs immutable?** Mutable can change in place (list, dict, set). Immutable cannot (int, str, tuple, frozenset).

**4. Shallow vs deep copy?** Shallow copies the outer object only, inner objects are shared. Deep copies everything recursively.

**5. What are `*args` and `**kwargs`?** Collect extra positional args into a tuple and extra keyword args into a dict.

**6. What is a decorator?** A function that takes a function and returns a wrapped function with extra behaviour.

**7. Generator vs iterator vs iterable?** Iterable can be looped. Iterator has `__next__`. Generator is an easy way to build an iterator with `yield`, and it is lazy.

**8. What is the GIL?** Global Interpreter Lock: only one thread runs Python bytecode at a time in CPython. Threads help with I/O-bound work, use `multiprocessing` for CPU-bound work.

**9. `__init__` vs `__new__`?** `__new__` creates the object, `__init__` initializes it.

**10. What is `self`?** Reference to the current instance, passed automatically to instance methods.

**11. `@staticmethod` vs `@classmethod`?** Static has no access to class or instance. Class method gets `cls`, often used as alternative constructor.

**12. What is MRO?** Method Resolution Order: the order Python searches classes for a method (C3 linearization). See `Class.__mro__`.

**13. How is memory managed?** Reference counting plus a cyclic garbage collector.

**14. Why avoid mutable default args?** Default is created once at definition time and shared across calls.

**15. `deepcopy` vs `copy`?** See Q4.

**16. List comprehension vs generator expression?** `[]` builds the whole list in memory, `()` is lazy.

**17. What is `__name__ == "__main__"`?** True when file is run directly, False when imported.

**18. Pass by value or reference?** "Pass by object reference": the function gets a reference to the same object. Mutating it affects the caller, reassigning the name does not.

**19. `append` vs `extend`?** `append` adds one item (even a list as a single item), `extend` adds each item of an iterable.

**20. What is a lambda?** Small anonymous single-expression function.

**21. Time complexity:** list index O(1), list `in` O(n), set/dict lookup O(1) average, list `insert(0)` O(n), `deque.appendleft` O(1), sort O(n log n).

**22. `try/except/else/finally`?** else runs if no exception, finally runs always.

**23. What are dunder methods?** Special methods like `__str__`, `__len__` that customise how objects behave with built-ins and operators.

**24. Multithreading vs multiprocessing vs asyncio?** Threads: I/O-bound, share memory. Processes: CPU-bound, separate memory. asyncio: many I/O tasks in one thread using `async/await`.

````python
# Quick async + threading taste (concept only, mid level)
import asyncio
from concurrent.futures import ThreadPoolExecutor

async def fetch(i):
    await asyncio.sleep(0.1)          # pretend network call
    return f"result {i}"

async def run_all():
    return await asyncio.gather(*(fetch(i) for i in range(3)))

try:
    print(asyncio.run(run_all()))
except RuntimeError:                   # inside Jupyter an event loop already runs
    print(await run_all())

with ThreadPoolExecutor(max_workers=3) as ex:
    print(list(ex.map(lambda n: n * n, range(5))))
````

````python
import pandas as pd

# we create a fake dataset so no download is needed
rng = np.random.default_rng(7)
n = 300
df = pd.DataFrame({
    "age": rng.integers(18, 60, n),
    "city": rng.choice(["Bangalore", "Mumbai", "Delhi", "Pune"], n),
    "dept": rng.choice(["IT", "HR", "Sales", "Ops"], n, p=[.4, .1, .3, .2]),
    "experience": rng.integers(0, 25, n),
})
df["salary"] = (25000 + df["experience"] * 3000 + rng.normal(0, 8000, n)).round(-2)
df.loc[rng.choice(n, 15, replace=False), "salary"] = np.nan          # inject missing values
df.loc[3, "salary"] = 900000                                         # inject an outlier
df = pd.concat([df, df.iloc[:5]], ignore_index=True)                 # inject duplicates

print(df.shape)
df.head()
````

````python
# STEP 2: look at the data
df.info()
print(df.describe())
print(df.describe(include="object"))
print("missing:\n", df.isna().sum())
print("duplicates:", df.duplicated().sum())
````

````python
# STEP 3: clean
df = df.drop_duplicates().reset_index(drop=True)
df["salary"] = df["salary"].fillna(df["salary"].median())      # median is safe against outliers

# outlier detection with IQR rule
q1, q3 = df["salary"].quantile([.25, .75])
iqr = q3 - q1
low, high = q1 - 1.5 * iqr, q3 + 1.5 * iqr
outliers = df[(df["salary"] < low) | (df["salary"] > high)]
print("outliers found:", len(outliers))
df_clean = df[df["salary"].between(low, high)].copy()
print(df_clean.shape)
````

````python
# STEP 5: groupby / bivariate analysis in tables
print(df_clean.groupby("dept")["salary"].agg(["count", "mean", "median", "max"]).round(0))
print(df_clean.pivot_table(values="salary", index="city", columns="dept", aggfunc="mean").round(0))
print(df_clean["city"].value_counts(normalize=True).round(2))
print(df_clean[["age", "experience", "salary"]].corr().round(2))

# useful pandas moves
df_clean["salary_band"] = pd.cut(df_clean["salary"], bins=[0, 40000, 70000, 1e9], labels=["low", "mid", "high"])
print(df_clean["salary_band"].value_counts())
print(df_clean.sort_values("salary", ascending=False).head(3))
print(df_clean.query("dept == 'IT' and experience > 10").shape)
````
