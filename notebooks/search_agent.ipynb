```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_ollama import ChatOllama
from langchain.agents import create_agent
from langchain_community.utilities import GoogleSerperAPIWrapper
from langgraph.checkpoint.memory import InMemorySaver
```

```python
search = GoogleSerperAPIWrapper()
```

```python
llm = ChatOllama(model="gemma4:e2b")
```

```python
memory = InMemorySaver()
```

```python
agent = create_agent(
    model=llm,
    tools=[search.run],
    system_prompt="You are an agent and you can use Google Search.",
    checkpointer=memory
)
```

```python
config = {
    "configurable": {
        "thread_id": "user-123"
    }
}
```

```python
question = "who won the ipl 2026"
```

```python
res = agent.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content": question
            }
        ]
    },
    config=config
)
```

```python
print(res["messages"][-1].content)
```

```python
# Follow-up question
res = agent.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content": "What were the runs of each team?"
            }
        ]
    },
    config=config
)

print(res["messages"][-1].content)
```
