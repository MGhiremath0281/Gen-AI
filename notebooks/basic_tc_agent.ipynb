```python
from langchain_ollama import ChatOllama
from langchain_core.tools import tool
```

```python
llm = ChatOllama(model="gemma4:e2b")
```

```python
@tool

def add_numbers(a:int , b:int):
    """
    It will return the sum of the two numbers

    Args:
        a: Number 1
        b: Number 3
    """
    return a+b

def mul_numbers(a:int , b:int):
    """
    It will return the multiple of the two numbers

    Args:
        a: Number 1
        b: Number 3
    """
    return a*b
```

```python
from langchain.agents import create_agent
```

```python
agent = create_agent(
    model=llm,
    tools= [add_numbers , mul_numbers ]
)
```
